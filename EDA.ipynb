```python
plt.figure(figsize=(10,5))
plt.hist(df['luas_tanaman'], bins=30)
plt.title("Distribusi Luas Tanaman")
plt.xlabel("Luas Tanaman  (ha)")
plt.ylabel("Frekuensi")
plt.show()
```

```python
plt.figure(figsize=(10,5))
plt.hist(df['luas_tanaman'], bins=30)
plt.title("Distribusi Luas Tanaman")
plt.xlabel("Luas Tanaman")
plt.ylabel("Frekuensi")
plt.show()
```

```python
df_group_tahun = df.groupby('tahun')['luas_tanaman'].sum().reset_index()

plt.figure(figsize=(10,5))
plt.plot(df_group_tahun['tahun'], df_group_tahun['luas_tanaman'], marker='o')
plt.title("Total Luas Tanaman Teh per Tahun")
plt.xlabel("Tahun")
plt.ylabel("Total Luas Tanaman (ha)")
plt.xticks(df_group_tahun['tahun'])
plt.grid(True)
plt.show()
```

```python
df_group_kab = df.groupby('nama_kabupaten_kota')['luas_tanaman'].sum().sort_values(ascending=False).head(10)

plt.figure(figsize=(12,6))
df_group_kab.plot(kind='bar')
plt.title("Top 10 Kabupaten/Kota dengan Total Luas Tanaman Terbesar")
plt.ylabel("Total Luas (ha)")
plt.xlabel("Kabupaten/Kota")
plt.xticks(rotation=45)
plt.show()
```

```python
df_group_kondisi = df.groupby('kondisi_tanaman')['luas_tanaman'].sum()

plt.figure(figsize=(8,5))
df_group_kondisi.plot(kind='bar')
plt.title("Total Luas Tanaman Berdasarkan Kondisi Tanaman")
plt.ylabel("Total Luas (ha)")
plt.xlabel("Kondisi Tanaman")
plt.xticks(rotation=0)
plt.show()
```
